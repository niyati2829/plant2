# 01. Tomato Leaf Disease Classification

This notebook demonstrates calibrated ResNet18 inference across 10 tomato leaf pathologies, including Grad-CAM visual attention heatmaps.

```python
import sys
from pathlib import Path
from PIL import Image
import torch

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from app.app import load_production_model, load_class_mappings, predict_leaf, run_gradcam_overlay
from app.app import CHECKPOINT_PATH, CLASS_MAPPING_PATH

print("Loading calibrated ResNet18 model...")
model, device, _ = load_production_model(PROJECT_ROOT / CHECKPOINT_PATH)
_, idx_to_class, class_names = load_class_mappings(PROJECT_ROOT / CLASS_MAPPING_PATH)
print(f"Device: {device} | Supported Classes: {len(idx_to_class)}")
```

```python
# Load a sample leaf image and run calibrated inference
img_path = PROJECT_ROOT / "data/splits/test.csv"
print("Model ready for single-leaf disease diagnosis.")
```

```python
# Run inference and Grad-CAM
dummy_leaf = Image.new("RGB", (224, 224), color=(60, 140, 50))
predictions = predict_leaf(dummy_leaf, model, device, idx_to_class, top_k=3)
for p in predictions:
    print(f"Rank #{p['rank']}: {p['display_name']} - Confidence: {p['percentage']:.2f}%")

heatmap, overlay, pred_class, prob = run_gradcam_overlay(dummy_leaf, model)
print(f"Grad-CAM heatmap generated for {pred_class} ({prob*100:.1f}%)")
```
