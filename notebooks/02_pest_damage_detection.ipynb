# 02. Tuta absoluta Pest Damage Detection & Segmentation

This notebook demonstrates foliar feeding-mine localization and instance segmentation using production YOLOv8n-Seg.

```python
import sys
from pathlib import Path
from PIL import Image

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from app.app import load_yolov8_detector, predict_pests_yolov8
from app.app import YOLO_CHECKPOINT_PATH

print("Loading YOLOv8n-Seg detector...")
detector = load_yolov8_detector(PROJECT_ROOT / YOLO_CHECKPOINT_PATH)
print("Detector loaded successfully!")
```

```python
# Run instance segmentation on foliage image
dummy_foliage = Image.new("RGB", (640, 640), color=(50, 130, 45))
annotated_img, detections, tuta_count, healthy_count = predict_pests_yolov8(
    dummy_foliage, detector, conf_threshold=0.25, iou_threshold=0.45
)
print(f"Detections -> Tuta Feeding Mines: {tuta_count} | Healthy Reference Foliage: {healthy_count}")
for det in detections:
    print(f"  Category: {det['category']} | Confidence: {det['confidence']:.2f} | BBox: {det['box']}")
```
